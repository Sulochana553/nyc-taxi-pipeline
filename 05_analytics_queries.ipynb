```python
%sql
-- Busiest pickup zones by hour
SELECT l.zone, t.hour, COUNT(*) AS trips
FROM workspace.default.fact_trips f
JOIN workspace.default.dim_location l ON f.pickup_location_key = l.location_key
JOIN workspace.default.dim_time t ON f.pickup_time_key = t.time_key
GROUP BY l.zone, t.hour
ORDER BY trips DESC
LIMIT 15;
```

```python
%sql
-- Average tip % by payment type and borough
SELECT p.description AS payment_type, l.borough,
       ROUND(AVG(f.tip_amount / NULLIF(f.fare_amount, 0)) * 100, 1) AS avg_tip_pct
FROM workspace.default.fact_trips f
JOIN workspace.default.dim_payment_type p ON f.payment_type_key = p.payment_type_key
JOIN workspace.default.dim_location l ON f.pickup_location_key = l.location_key
GROUP BY p.description, l.borough
ORDER BY avg_tip_pct DESC;
```

```python
%sql
-- Weekday vs weekend trip patterns
SELECT d.is_weekend, t.time_of_day_bucket, COUNT(*) AS trips, ROUND(AVG(f.total_amount), 2) AS avg_fare
FROM workspace.default.fact_trips f
JOIN workspace.default.dim_date d ON f.pickup_date_key = d.date_key
JOIN workspace.default.dim_time t ON f.pickup_time_key = t.time_key
GROUP BY d.is_weekend, t.time_of_day_bucket
ORDER BY d.is_weekend, trips DESC;
```
