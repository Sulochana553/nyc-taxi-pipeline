```python
import urllib.request
import os
import time

dbutils.widgets.text("year_month", "2024-01", "Year-Month to extract (YYYY-MM)")
year_month = dbutils.widgets.get("year_month")

VOLUME_PATH = "/Volumes/workspace/default/raw_files"
TRIPS_URL = f"https://d37ci6vzurychx.cloudfront.net/trip-data/yellow_tripdata_{year_month}.parquet"
ZONE_URL = "https://d37ci6vzurychx.cloudfront.net/misc/taxi_zone_lookup.csv"

def log(msg):
    ts = time.strftime("%Y-%m-%d %H:%M:%S")
    print(f"[{ts}] {msg}")

def extract_and_land(url, dest_path, label):
    log(f"START extract: {label} from {url}")
    try:
        urllib.request.urlretrieve(url, dest_path)
        size_mb = os.path.getsize(dest_path) / (1024 * 1024)
        log(f"SUCCESS: {label} landed at {dest_path} ({size_mb:.1f} MB)")
        return True
    except Exception as e:
        log(f"FAILED: {label} — {e}")
        raise
```

```python
trips_dest = f"{VOLUME_PATH}/yellow_tripdata_{year_month}.parquet"
zone_dest = f"{VOLUME_PATH}/taxi_zone_lookup.csv"

extract_and_land(TRIPS_URL, trips_dest, "yellow taxi trips")
extract_and_land(ZONE_URL, zone_dest, "taxi zone lookup")

log("Extract and land complete.")
```
